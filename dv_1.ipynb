```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Name : Aditya Mahadev Kumbhar")
print("612415094")
print("  \n")

# Load Iris dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
df = pd.read_csv(url)


print(df)
```

```python
print("\nColumn Names:")
print(df.columns)
```

```python
print("\nStatistical Summary:")
print(df.describe())
```

```python
avg_petal = df.groupby("species")["petal_length"].mean()

avg_petal.plot(kind="bar")

plt.title("Average Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Average Petal Length")
plt.show()
```

```python
plt.plot(df.index[:40], df["sepal_length"][:40])

plt.title("Sepal Length for First 40 Iris Samples")
plt.xlabel("Sample Index")
plt.ylabel("Sepal Length")
plt.show()
```

```python
plt.hist(df["sepal_length"], bins=6)

plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.show()
```
